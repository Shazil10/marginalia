# Agent 2 Exploratory Backtest

This notebook runs one real ingested paper from the local synthetic SQLite database and inspects the resulting tear sheet and equity curve artifacts.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

SRC = ROOT / "src"
if str(SRC) not in sys.path:
    sys.path.append(str(SRC))

ROOT
```

```python
from agent2.agents.orchestrator import OrchestratorAgent
from agent2.ingest import load_agent1_from_db

paper = load_agent1_from_db(ROOT / "data" / "papers.sqlite", "tactical")
paper.title, paper.paper_category, paper.asset_universe_type
```

```python
orchestrator = OrchestratorAgent()
result = orchestrator.run(
    paper,
    start=__import__("datetime").date(2010, 1, 1),
    end=__import__("datetime").date(2024, 12, 31),
    artifacts_dir=ROOT / "artifacts",
    run_id="notebook-tactical",
)
result.output_dir
```

```python
result.result.tear_sheet.model_dump()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

equity = pd.read_csv(result.output_dir / "equity_curve.csv", index_col=0, parse_dates=True)
equity[["gross_equity", "net_equity", "benchmark_equity"]].plot(figsize=(11, 5), title="Tactical Paper Backtest")
plt.ylabel("Growth of $1")
plt.grid(alpha=0.25)
plt.show()
```
